```python
import random

seq_length = 2500
dna_sequence = ''.join(random.choices('ATCG', k=seq_length))

samples = []
for _ in range(10):
    sample_len = random.randint(100, seq_length)
    start_idx = random.randint(0, seq_length - sample_len)
    sample = dna_sequence[start_idx : start_idx + sample_len]
    samples.append(sample)

sorted_samples = sorted(samples, key=len, reverse=True)

for i, sample in enumerate(sorted_samples):
    print(f"Band {i+1}: {len(sample):4d} bp  |  Seq: {sample[:10]}...{sample[-10:]}")
```

```python
import random
import re

seq_length = 2500
dna_sequence = ''.join(random.choices('ATCG', k=seq_length))

injected_unit = "ATGC"
injected_str = injected_unit * 8
dna_sequence = dna_sequence[:1200] + injected_str + dna_sequence[1200 + len(injected_str):]

def detect_repetitions(sequence):
    pattern = re.compile(r'(([ATCG]{3,5})\2{5,9})')
    matches = pattern.finditer(sequence)

    results = []
    for match in matches:
        full_match = match.group(1)
        unit = match.group(2)
        repetitions = len(full_match) // len(unit)
        results.append({
            "start": match.start(),
            "end": match.end(),
            "unit": unit,
            "repetitions": repetitions,
            "sequence": full_match
        })
    return results

found_repeats = detect_repetitions(dna_sequence)

if found_repeats:
    for idx, repeat in enumerate(found_repeats, 1):
        print(f"Match #{idx}")
        print(f"Position:   {repeat['start']} to {repeat['end']}")
        print(f"Unit:       '{repeat['unit']}' (Length: {len(repeat['unit'])} bases)")
        print(f"Repeats:    {repeat['repetitions']} times")
        print(f"Sequence:   {repeat['sequence']}\n")
else:
    print("No repetitions matching the criteria were found.")
```
